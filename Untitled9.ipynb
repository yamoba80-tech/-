```python
# 1. تثبيت المكتبات المطلوبة
!pip install -q diffusers transformers accelerate safetensors invisible-watermark gradio

import torch
import gradio as gr
from diffusers import DiffusionPipeline

# 2. تحميل النموذج مباشرة (بدون أي رمز تسجيل)
print("⏳ جاري تحميل النموذج... قد يستغرق بضع دقائق")
pipe = DiffusionPipeline.from_pretrained(
    "stabilityai/stable-diffusion-xl-base-1.0",
    torch_dtype=torch.float16,
    use_safetensors=True,
    variant="fp16"
)
# تفريغ الذاكرة لضمان العمل على كرت T4 المجاني
pipe.enable_model_cpu_offload()
print("✅ تم التحميل بنجاح! الواجهة جاهزة.")

# 3. دالة توليد الصورة
def generate_image(prompt, negative_prompt, steps):
    image = pipe(
        prompt=prompt,
        negative_prompt=negative_prompt,
        num_inference_steps=int(steps),
        guidance_scale=7.5
    ).images[0]
    return image

# 4. بناء واجهة الويب التفاعلية
with gr.Blocks(title="SDXL Generator") as demo:
    gr.Markdown("## 🎨 مولّد صور الذكاء الاصطناعي (بدون تسجيل)")
    with gr.Row():
        with gr.Column():
            prompt = gr.Textbox(label="الوصف (بالإنجليزية لأفضل نتيجة)", value="A majestic lion jumping from a big stone at night, highly detailed")
            negative_prompt = gr.Textbox(label="ما لا تريده", value="blurry, low quality, distorted")
            steps = gr.Slider(minimum=10, maximum=50, value=30, step=5, label="عدد الخطوات")
            btn = gr.Button("🎨 توليد الصورة", variant="primary")
        with gr.Column():
            output = gr.Image(label="الصورة الناتجة")

    btn.click(fn=generate_image, inputs=[prompt, negative_prompt, steps], outputs=output)

    # 5. تشغيل الواجهة
    demo.launch(share=True)
```
